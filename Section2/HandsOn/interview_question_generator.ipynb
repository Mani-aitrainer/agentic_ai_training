```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
```

```python
os.environ["OPENAI_API_KEY"] = ""
```

```python
prompt_template = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are an expert technical interviewer and HR panelist with 10 years 
of experience designing interview questions for hiring processes across industries. 
Your questions are clear, relevant, and appropriately challenging for the stated 
difficulty level. Always write in a professional, neutral tone."""
    ),
    (
        "human",
        """Generate a set of interview questions for a hiring panel.

Job Role: {job_role}
Skill Area to Focus On: {skill_area}
Difficulty Level: {difficulty}

Requirements:
- Generate 5 questions total
- Order them from easiest to hardest
- For each question, add a one-line note on what a good answer should cover
- Avoid yes/no questions — favor questions that reveal reasoning and experience
- Use clear, simple language"""
    )
])
```

```python
llm = ChatOpenAI(
    model="gpt-3.5-turbo",   # The GPT model version to use
    temperature=0.7           # Creativity level: 0 = robotic, 1 = very creative
)
```

```python
output_parser = StrOutputParser()
```

```python
chain = prompt_template | llm | output_parser
```

```python
role_input = {
    "job_role"   : "Data Scientist",
    "skill_area" : "Python, SQL, and statistical modeling",
    "difficulty" : "Intermediate"
}
```

```python
questions = chain.invoke(role_input)

print("✅ Interview Questions Generated!")
print("=" * 55)
print()
print(questions)
print()
print("=" * 55)
```

```python
test_role = {
    "job_role"   : "Frontend Developer",
    "skill_area" : "React, state management, and UI performance",
    "difficulty" : "Intermediate"
}

temperatures = [0.1, 0.9]
```

```python
for temp in temperatures:
    print(f"{'='*55}")
    print(f"🌡️  Temperature = {temp}  ({'Conservative' if temp < 0.5 else 'Creative'})")
    print(f"{'='*55}")
    
    # Create a new model instance with this temperature
    temp_llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=temp)
    temp_chain = prompt_template | temp_llm | output_parser
    
    result = temp_chain.invoke(test_role)
    print(result)

print(f"{'='*55}")
print("💡 Notice the difference in phrasing and question variety between the two!")
print(f"{'='*55}")
```
